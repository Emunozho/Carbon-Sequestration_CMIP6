```python
import numpy as np
import matplotlib.ticker as tkr
import xarray as xr
import matplotlib.pyplot as plt
import cartopy.crs as ccrs
from matplotlib import colors
```

## read saved files

```python
wnep=xr.open_dataset('~/ResultsCS/Weights/nepTRENDY_ob_FLUXCOM-X.nc') # weights CMIP6-FLUXCOM-X.
wnbp=xr.open_dataset('~/ResultsCS/Weights/nbpTRENDY_ob_CarboScope.nc')  # weights CMIP6-CarboScope
vnep=xr.open_dataset('~/ResultsCS/Weights/nepTRENDY_ob_FLUXCOM-X_variance.nc') # variance CMIP6-FLUXCOM-X
vnbp=xr.open_dataset('~/ResultsCS/Weights/nbpTRENDY_ob_CarboScope_variance.nc')  # variance CMIP6-CarboScope
```

## TRENDY-FLUXCOM-X

```python
plt.rcParams.update({'font.size':8})
fig,ax=plt.subplots(ncols=4,nrows=5,figsize=[12,10],subplot_kw={"projection": ccrs.Robinson()})
#fig.suptitle('TRENDY-FLUXCOM-X (NEP)')     
xxl=list(wnep.keys()); xxw=[]    
a=-1
levels=np.linspace(0,1,(len(xxl))*2+1)
divnorm=colors.TwoSlopeNorm(vmin=0.,vcenter=1/len(xxl),vmax=1)
for j in range(len(xxl)):
#for j in range(1):
    if j/4==int(j/4):a=a+1
    xxv=wnep[xxl[j]]
    pt=ax[a,j-4*a].contourf(xxv.lon,xxv.lat,xxv,transform=ccrs.PlateCarree(),cmap='seismic',levels=levels,norm=divnorm)
    ax[a,j-4*a].coastlines()
    ax[a,j-4*a].set_title(xxl[j])
    cbar=plt.colorbar(pt,shrink=0.8,orientation='horizontal',ax=ax[a,j-4*a],pad=0.04,aspect=30,format=tkr.FormatStrFormatter('%.1g'))
    for c in pt.collections:c.set_edgecolor("face")
fig.savefig('/home/_ehoyos/cmip6stop/CS/Figures/Weights/TRENDY_FLUXCOM_weights.pdf')
```

## TRENDY-CarboScope

```python
plt.rcParams.update({'font.size':8})
fig,ax=plt.subplots(ncols=3,nrows=6,figsize=[10,12],subplot_kw={"projection": ccrs.Robinson()})
#fig.suptitle('TRENDY-CarboScope (NBP)')     
xxl=list(wnbp.keys()); xxw=[]    
a=-1
levels=np.linspace(0,1,(len(xxl))*2+1)
divnorm=colors.TwoSlopeNorm(vmin=0.,vcenter=1/len(xxl),vmax=1)
for j in range(len(xxl)):
#for j in range(1):
    if j/3==int(j/3):a=a+1
    xxv=wnbp[xxl[j]]
    pt=ax[a,j-3*a].contourf(xxv.lon,xxv.lat,xxv,transform=ccrs.PlateCarree(),cmap="seismic",levels=levels,norm=divnorm)
    ax[a,j-3*a].coastlines()
    ax[a,j-3*a].set_title(xxl[j])
    cbar=plt.colorbar(pt,shrink=0.8,orientation='horizontal',ax=ax[a,j-3*a],pad=0.04,aspect=30,format=tkr.FormatStrFormatter('%.1g'))
    for c in pt.collections:c.set_edgecolor("face")
fig.savefig('/home/_ehoyos/cmip6stop/CS/Figures/Weights/TRENDY_CarboScope_weights.pdf')
```
